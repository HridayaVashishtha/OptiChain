```python
import pandas as pd

sales = pd.read_csv("../data/raw/fact_sales_dirty.csv")

clean = sales.copy()
clean = clean.drop_duplicates(subset=["date","store_id","product_id","transaction_id"])
clean = clean[clean["selling_price"] > 0]
clean = clean[clean["quantity"] >= 0]
clean["discount_pct"] = clean["discount_pct"].fillna(0)
clean.to_csv("../data/processed/fact_sales_clean.csv", index=False)
```
